```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
with open("data/BRFSS.csv", "r", encoding="utf-8") as f:
    data_str = f.read()

# some of the feature are wrongly formatted
# we start by removing all the comma (,) and quotes (") from the text features

data_str = data_str.replace("No, pre-diabetes or borderline diabetes", "No pre-diabetes or borderline diabetes")
data_str = data_str.replace("Yes, but female told only during pregnancy", "Yes but female told only during pregnancy")
data_str = data_str.replace('"', '')

with open('data/BRFSS2.csv', 'w', encoding='utf-8') as f:
    f.write(data_str)
```

```python
data = pd.read_csv("data/BRFSS2.csv")
data
```

Now we check each feature for preprocessing

```python
# General_Health 
feat_name = "General_Health"

data["General_Health"].value_counts(dropna=False).plot(kind='bar')
plt.show()

# the feature General_Health is categorical with 'Excellent', 'Very Good', 'Good', 'Fair' and 'Poor' as possible values
# we map them in integer keeping the order Poor=1 < Fair < ... < Excellent=5

# moreover there are no missing value

map = {
    'Excellent': 5,
    'Very Good': 4,
    'Good': 3,
    'Fair': 2,
    'Poor': 1
}

data["General_Health"] = data["General_Health"].map(map)
```

```python
# Checkup
data["Checkup"].value_counts(dropna=False).plot(kind="bar")
plt.show()

# the feature Checkup is categorical with 'Never', '5 or more years ago', 'Within the past 5 years', 'Within the past 2 years' and 'Within the past year' as possible values
# we map them in integer keeping the order Never=1 < 5 or more years ago < ... < Within the past year=5

# moreover there are no missing value

map = {
    "Within the past year": 5,
    "Within the past 2 years": 4,
    "Within the past 5 years": 3,
    "5 or more years ago": 2,
    "Never": 1
}

data["Checkup"] = data["Checkup"].map(map)
```

```python
# Exercise
data["Exercise"].value_counts(dropna=False).plot(kind="bar")
plt.show()

# the feature Excercise is boolean with 'Yes' and 'No' as possible values
# we map them in integer 0/1

# moreover there are no missing value

map = {
    "Yes": 1,
    "No": 0
}

data["Exercise"] = data["Exercise"].map(map)
```
